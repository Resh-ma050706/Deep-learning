```python
# TASK A: DATASET PREPARATION

import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
# Student Details
student_name = "S.RESHMA"
roll_no = "24BAD097"
print("Student Name :", student_name)
print("Roll Number  :", roll_no)
print("Task A: Dataset Preparation")
# Load CIFAR-10 dataset
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.cifar10.load_data()
# Class names
class_names = [
    "Airplane", "Automobile", "Bird", "Cat", "Deer",
    "Dog", "Frog", "Horse", "Ship", "Truck"
]
# Normalize images
x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0
# Create validation set
x_val = x_train[-10000:]
y_val = y_train[-10000:]
x_train = x_train[:-10000]
y_train = y_train[:-10000]
print("\nDataset Information")
print("Training images   :", x_train.shape)
print("Validation images :", x_val.shape)
print("Testing images    :", x_test.shape)
# Display sample images
plt.figure(figsize=(10, 6))
for i in range(10):
    plt.subplot(2, 5, i + 1)
    plt.imshow(x_train[i])
    plt.title(class_names[y_train[i][0]])
    plt.axis("off")
plt.tight_layout()
plt.show()
```

```python
# TASK B: CNN MODEL IMPLEMENTATION

import tensorflow as tf
# Student Details
student_name = "S.RESHMA"
roll_no = "24BAD097"
print("Student Name :", student_name)
print("Roll Number  :", roll_no)
print("Task B: CNN Model Implementation")
# CNN Model
model = tf.keras.Sequential([
    # Convolution Layer 1
    tf.keras.layers.Conv2D(
        32, (3, 3), activation="relu",
        input_shape=(32, 32, 3)
    ),
    # Max Pooling
    tf.keras.layers.MaxPooling2D((2, 2)),
    # Convolution Layer 2
    tf.keras.layers.Conv2D(
        64, (3, 3), activation="relu"
    ),
    # Max Pooling
    tf.keras.layers.MaxPooling2D((2, 2)),
    # Convolution Layer 3
    tf.keras.layers.Conv2D(
        64, (3, 3), activation="relu"
    ),
    # Flatten
    tf.keras.layers.Flatten(),
    # Fully Connected Layer
    tf.keras.layers.Dense(64, activation="relu"),
    # Softmax Output Layer
    tf.keras.layers.Dense(10, activation="softmax")
])
# Compile model
model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)
# Display model architecture
model.summary()
```

```python
# TASK C: MODEL TRAINING AND EVALUATION

import matplotlib.pyplot as plt
print("Student Name :", "S.RESHMA")
print("Roll Number  :", "24BAD097")
print("Task C: Model Training and Evaluation")
# Train the CNN
history = model.fit(
    x_train,
    y_train,
    epochs=10,
    batch_size=64,
    validation_data=(x_val, y_val),
    verbose=1
)
# Evaluate on test data
test_loss, test_accuracy = model.evaluate(
    x_test,
    y_test,
    verbose=0
)
# Final results
print("\nFINAL RESULTS")
print("-" * 40)
print("Training Accuracy   :",
      round(history.history["accuracy"][-1], 4))
print("Validation Accuracy :",
      round(history.history["val_accuracy"][-1], 4))
print("Training Loss       :",
      round(history.history["loss"][-1], 4))
print("Validation Loss     :",
      round(history.history["val_loss"][-1], 4))
print("Testing Accuracy    :",
      round(test_accuracy, 4))
# Accuracy vs Epoch
plt.figure(figsize=(8, 5))
plt.plot(
    history.history["accuracy"],
    label="Training Accuracy"
)
plt.plot(
    history.history["val_accuracy"],
    label="Validation Accuracy"
)
plt.title("Accuracy vs Epoch")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()
plt.grid(True)
plt.show()
# Loss vs Epoch
plt.figure(figsize=(8, 5))
plt.plot(
    history.history["loss"],
    label="Training Loss"
)
plt.plot(
    history.history["val_loss"],
    label="Validation Loss"
)
plt.title("Loss vs Epoch")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.grid(True)
plt.show()
```

```python
# TASK D: PERFORMANCE ANALYSIS

import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

print("Student Name :", "S.RESHMA")
print("Roll Number  :", "24BAD097")
print("Task D: Performance Analysis")
# Make predictions
predictions = model.predict(x_test, verbose=0)
predicted_labels = np.argmax(predictions, axis=1)
actual_labels = y_test.flatten()
# Classification accuracy
accuracy = np.mean(predicted_labels == actual_labels)
print("\nClassification Accuracy:",
      round(accuracy * 100, 2), "%")

# -----------------------------
# CONFUSION MATRIX
# -----------------------------
cm = confusion_matrix(
    actual_labels,
    predicted_labels
)
plt.figure(figsize=(10, 8))
disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=class_names
)
disp.plot(
    xticks_rotation=45,
    cmap="Blues"
)
plt.title("CIFAR-10 Confusion Matrix")
plt.show()
# -----------------------------
# SAMPLE PREDICTIONS
# -----------------------------
plt.figure(figsize=(12, 6))
for i in range(10):
    plt.subplot(2, 5, i + 1)
    plt.imshow(x_test[i])
    actual = class_names[actual_labels[i]]
    predicted = class_names[predicted_labels[i]]
    plt.title(
        "Actual: " + actual +
        "\nPredicted: " + predicted
    )
    plt.axis("off")
plt.tight_layout()
plt.show()
# -----------------------------
# MISCLASSIFIED IMAGES
# -----------------------------

wrong_indices = np.where(
    predicted_labels != actual_labels
)[0]

print("Number of misclassified images:",
      len(wrong_indices))
plt.figure(figsize=(12, 6))
for i in range(min(10, len(wrong_indices))):
    index = wrong_indices[i]
    plt.subplot(2, 5, i + 1)
    plt.imshow(x_test[index])
    actual = class_names[actual_labels[index]]
    predicted = class_names[predicted_labels[index]]
    plt.title(
        "Actual: " + actual +
        "\nPredicted: " + predicted
    )
    plt.axis("off")
plt.tight_layout()
plt.show()
```
